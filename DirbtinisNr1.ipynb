```python
import numpy as np

def sigmoidine(x):
  return round(1/(1 + np.exp(-x)))

def slenkstine(x):
  if x >=0:
    return 1
  return 0

inputs = np.array([[-0.2, 0.5], [0.2, -0.7], [0.8, -0.8], [0.8, 1]])
inputs

outputs = np.array([0,0,1,1])
outputs
```

```python
import random
def random_search():
    while True:

        w0 = round(random.uniform(-5, 5), 3)
        w1 = round(random.uniform(-5, 5), 3)
        w2 = round(random.uniform(-5, 5), 3)
        correct_classes = True

        for i in range(len(inputs)):
            weighted_sum = w0 + w1 * inputs[i][0] + w2 * inputs[i][1]
            prediction = slenkstine(weighted_sum)
            if prediction != outputs[i]:
              correct_classes = False
              break
        if correct_classes == True:
              return w0, w1, w2

best_weights_random = random_search()
print("Geriausi svoriai:", best_weights_random)
```

```python
def pasitikrinimas(w0, w1, w2):
    for i in range(len(inputs)):
        weighted_sum = w0 + w1 * inputs[i][0] + w2 * inputs[i][1]
        prediction = slenkstine(weighted_sum)
        print(f"Input: {inputs[i]}, Prediction: {prediction}, Expected: {outputs[i]}")
```

```python
def perrinkimas():
    for w0 in np.arange(-5, 5, 0.2):
        for w1 in np.arange(-5, 5, 0.2):
            for w2 in np.arange(-5, 5, 0.2):
                correct_classes = True

                for i in range(len(inputs)):
                    weighted_sum = w0 + w1 * inputs[i][0] + w2 * inputs[i][1]
                    prediction = slenkstine(weighted_sum)


                    if prediction != outputs[i]:
                        correct_classes = False
                        break

                if correct_classes == True:
                    return w0, w1, w2

best_weights_perrinkimas = perrinkimas()
print("Geriausi svoriai:", best_weights_perrinkimas)
```

```python
print("Pasitikrinimas su gautais random svoriais")
pasitikrinimas(best_weights_random[0], best_weights_random[1], best_weights_random[2])
print("\nPasitikrinimas su gautais perrinktinais svoriais \n")
pasitikrinimas(best_weights_perrinkimas[0], best_weights_perrinkimas[1], best_weights_perrinkimas[2])
```
